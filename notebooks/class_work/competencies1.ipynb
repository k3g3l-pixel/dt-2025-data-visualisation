```python
import pandas as pd
df=pd.read_csv("Competencies1.csv")
df.head() 
```

```python
import matplotlib.pyplot as plt
avg_hours=df.groupby("Student")["Hours_Studied"].mean()
avg_hours.plot(kind="bar", x="Student", y="Hours_Studied")
plt.title("Who puts in the most effort")
plt.show()
```

who performs best

```python
avg_grades=df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
avg_grades.plot(kind="bar", x="Student", y="Grade", legend=False)
plt.title("Who puts in the most effort")
plt.ylabel("Grade")
plt.show()
```

```python
avg_grades=df.groupby("Student")["Grade"].mean().sort_values(ascending=False)
plt.figure(figsize=(9, 9))
avg_grades.plot(kind="pie", x="Student", y="Grade", legend=False, autopct='%1.1f%%')

plt.title("Who puts in the most effort")
plt.ylabel("Grade")
plt.show()
```

more study=better grades

```python
plt.scatter(df["Hours_Studied"], df["Grade"], s=df["Hours_Studied"]*30, alpha=0.6)
plt.title("Study=Grades")
plt.xlabel("Hours")
plt.ylabel("Grades")
plt.show()
```

```python
hours_min = df["Hours_Studied"].min()
hours_q1 = df["Hours_Studied"].quantile(0.25)
print("Minimum hours:", hours_min, "Q1:", hours_q1)
plt.figure(figsize=(8, 6))
df[["Hours_Studied", "Grade"]].boxplot()
plt.ylabel("Value")
plt.show()
```
