```python
import pandas as pd
df=pd.read_csv("sports.csv")
df.head() #
```

```python
df.describe()
```

box plot for each athlete to find nr of years of practice

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.ylabel("Years")
plt=title("Years of practice per athlete")
plt.show
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age")
plt.show
```

age vs practice years

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*30, alpha=0.6)
plt.xlabel("Age")
plt.ylabel("Practice years")
plt.title("Age vs practice years")
plt.show
```

avg age per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.title("avg age per sport")
plt.ylabel("avg_age")
plt.show
```
