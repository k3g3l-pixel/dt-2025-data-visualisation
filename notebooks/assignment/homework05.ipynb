now using pandas we'll see the dataset and using 'head' we'll see only the first few rows

```python
import pandas as pd
df=pd.read_csv("Homework05sheet.csv")
df.head()
```

using matplotlib we can see the data in different charts
in this bar chart we can see the avergae duration of each movie, from longest to shortest

```python
import matplotlib.pyplot as plt
avg_duration=df.groupby("Movie_title")["Duration"].mean().sort_values(ascending=False)
avg_duration.plot(kind="bar", x="Movie_title", y="Duration", legend=False)
plt.ylabel("Duration")
plt.xlabel("Movie title")
plt.title("Duration of each movie")
plt.show()
```

the longest movie here is the LOTR: the Fellowship of the Ring, while the shortest is The Cat Returns
movies on average seem to be an hour and a half long, and while there are those that exceed the 2 hour mark, the closer they get to the 3 hour mark the fewer of them there are

next we see a pie chart about movies and their ratings in percentiles

```python
avg_rating=df.groupby("Movie_title")["Rating"].mean().sort_values(ascending=False)
plt.figure(figsize=(9, 9))
avg_rating.plot(kind="pie", x="Movie_title", y="Rating", legend=False, autopct='%1.1f%%')
plt.title("What movie has the best rating?")
plt.ylabel("Rating")
plt.show
```

the movie with the best rating is POTC: Curse of the Black Pearl at 5.4%
while Snow White has the worst at 1.1%
most movie ratings are at 4.3%, with only a few exceeding it or being less than it

next we see a scatterplot abou box offices and ratings

```python
plt.scatter(df["Box_office"], df["Rating"], s=df["Box_office"]*30, alpha=0.3)
plt.title("Does doing well in the box office mean getting a good rating?")
plt.ylabel("Rating")
plt.xlabel("Box office")
plt.show()
```

While doing well in the box office does increase a films chances of getting a good rating
we can see that plenty of movies with small box offices got high ratings, from 7 to 9
and yet it still shows that high box office does equal a high rating
the exception being 1 movie that scored a rating of 2 despite its 200 thousand box office

In this last one we see a boxplot about durations and box offices

```python
duration_min=df["Duration"].min()
duration_q1=df["Duration"].quantile(0.25)
duration_max=df["Duration"].max()
duration_q3=df["Duration"].quantile(0.75)
print("Minimum duration:", duration_min, "Q1:", duration_q1, "Maximum duration:", duration_max, "Q3:", duration_q3)

boxoffice_min=df["Box_office"].min()
boxoffice_q1=df["Box_office"].quantile(0.25)
boxoffice_max=df["Box_office"].max()
boxoffice_q3=df["Box_office"].quantile(0.75)
print("Minimum box office:", boxoffice_min, "Q1:", boxoffice_q1, "Maximum box office:", boxoffice_max, "Q3:", boxoffice_q3)

plt.figure(figsize=(8,6))
df[["Duration", "Box_office"]].boxplot()
plt.title("How consistent are durations and box offices")
plt.ylabel("Value")
plt.show
```

the consistency between durations is, on the grand scale, pretty consistent, when compared to box offices.
box offices, on the other hand, are very incosistent, ranging from 1 thousand to over a billion,
yet a majority of them are between 153 thound and 775 thousand.
durations are predominantly between 1 hour and 40 minutes to 2 hours and 25 minutes, both containing exceptions.
the minimum of durations is 75 minutes, while the max being 178 minutes, nearly 3 hours long.
with box offices the min was 1 thousand, while the max was 1 billion 66 thousand.
