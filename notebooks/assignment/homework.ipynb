```python
import pandas as pd
df=pd.read_csv("Homework.csv")
df
```

```python

import matplotlib.pyplot as plt
df.plot(kind="bar", x="Genre", y="Duration", legend=False)
#plt.ylabel("Years")
#plt=title("Years of practice per athlete")
plt.show
```

```python
avg_duration=df.groupby("Genre")["Duration"]
avg_duration.plot(kind="bar", x="Movie title ", y="Duration")
plt.title("What movie has the longest duration?")
plt.show()
```
