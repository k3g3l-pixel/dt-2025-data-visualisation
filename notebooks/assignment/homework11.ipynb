Homework 11: Automating text processing and word cloud generation in python

```python
sentences="Olive oil obtained from olives harvested when ripe. Generally speaking, it is mildly fruity, with sweet notes and only a hint of bitterness and spiciness. It is the mildest, sweetest and least spicy olive oil. It pairs well with mildly flavoured soups, boiled vegetables with subtle flavours, eggs, mushrooms, dainty desserts, light cold salads, tender fish, mild sauces, boiled potatoes and chestnuts." #text that will be used 
words=sentences.split(" ") #no longer a string of words and spaces. Each word stands alone now
print(words) #prints each individual word while ignoring spaces
print("number of words:",len(words)) #to know the exact amount of words in the text
print(*words[-15:]) #to see the last 15 words
print(*words[::4]) #to see every 4th word
print(*words[8])#to see the 9th word
```
