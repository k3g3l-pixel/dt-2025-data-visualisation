```python
sentences="Olive oil obtained from olives harvested when ripe. Generally speaking, it is mildly fruity, with sweet notes and only a hint of bitterness and spiciness. It is the mildest, sweetest and least spicy olive oil. It pairs well with mildly flavoured soups, boiled vegetables with subtle flavours, eggs, mushrooms, dainty desserts, light cold salads, tender fish, mild sauces, boiled potatoes and chestnuts." #text that will be used 
words=sentences.split(" ") #no longer a string of words and spaces. Each word stands alone now
print(words) #prints each individual word while ignoring spaces
print("number of words:",len(words)) #to know the exact amount of words in the text
print(*words[-15:]) #to see the last 15 words
print(*words[::4]) #to see every 4th word
print(*words[8])#to see the 9th word
```

```python
import urllib.request #lets me access the URL
text=urllib.request.urlopen("https://raw.githubusercontent.com/k3g3l-pixel/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/assignment/oliveoil11.txt").read().decode("utf-8") #opens the URL, reads and decodes text from bytes into a string
print(text[0:]) #to check that the text is actually going to be displayed and that previous code isnt faulty
```

```python
from wordcloud import WordCloud #can use wordcloud now
WordCloud().generate_from_text(text).to_image() #generates a wordcloud from previous text
```

```python
WordCloud(stopwords=["from","harvested","and","it","is","the","of","when","It","obtained","Generally","speaking","with","only","hint","pairs","well","a","notes"],background_color="pink").generate_from_text(text).to_image()
#these words will not appeared in the wordcloud. Shows descriptive words in the text as well as nouns
#ive changed the background color to pink
#"oil" and "olive" are the largest because they appear the most in the text, along with "boiled" and "mildly"
```

The words that pop out the most show that the text is of something mild, with perhaps no specific taste. Yet the smaller words reaveal there is some taste, like "sweet" and "spiciness". Theres something of a contradiction, making one interested in reading further the full text.
