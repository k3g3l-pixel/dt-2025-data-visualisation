```python
import pandas as pd
df=pd.read_csv("homework_05_data.csv")
df.head()
```

here we can see the titles, genres, years, ratings, box offices and durations of the first few rows of movies

```python
import matplotlib.pyplot as plt
avg_duration=df.groupby("Movie_title")["Duration"]
avg_duration.plot(kind="bar", x="Movie_title", y="Duration")
plt.title("What movie has the longest duration?")
plt.show()
```

Here we can see the movies 

```python
avg_ratings=df.groupby("Movie_title")["Rating"].mean().sort_values(ascending=False)
plt.figure(figsize=(9, 9))
avg_ratings.plot(kind="pie", x="Movie_title", y="Rating", legend=False, autopct='%1.1f%%')

plt.title("What movie has the best rating")
plt.ylabel("Rating")
plt.show()
```

Here we can see

```python
plt.scatter(df["Box_office"], df["Rating"], s=df["Box_office"]*30, alpha=0.6)
plt.title("Does a good Rating mean doing well in the box office?")
plt.xlabel("Box_office")
plt.ylabel("Rating")
plt.show()
```

Here we can see

```python
duration_min = df["Duration"].min()
duration_min_q1 = df["Duration"].quantile(0.25)
print("Minimum duration:", duration_min, "Q1:", duration_q1)
plt.figure(figsize=(8, 6))
df[["Duration", "Rating"]].boxplot()
plt.ylabel("Value")
plt.show()
```
