```python
import pandas as pd
df=pd.read_csv("homework5data.csv")
df.head()
```

```python
import matplotlib.pyplot as plt
avg_duration=df.groupby("Genre")["Box_office"]
avg_duration.plot(kind="bar", x="Genre", y="Box_office")
plt.title("What movie has the longest duration?")

plt.show
```
